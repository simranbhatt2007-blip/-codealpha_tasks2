```python
# Lecture 1 - Example1 - A structured record as a python dictionary
# Create a dictionary to store student information
student = {
    "Name": "Amit",
    "Age": 21,
    "Marks": 85
}

# Print the complete dictionary
print("Student Details:")
print(student)

# Print the data type of the variable
print("Data Type:")
print(type(student))
# see the output below
```

```python
# Create a dictionary with college details
# Example-2 A Semi -Structured Record with Nested Data
college = {
    "name": "ABC Institute",
    "departments": [
        {"name": "CSE", "students": 120},
        {"name": "ECE", "students": 90}
    ]
}

# Print the name of the first department
print("First Department:")
print(college["departments"][0]["name"])
# See the otput below
```

```python
# Store the review in a string
# Example-3 Representing Unstructured Text
review = "The product quality was excellent, but delivery was delayed by two days."

# Split the review into individual words
words = review.split()

# Print the total number of words
print("Word Count:", len(words))

# Check if the word "delayed" is present in the review
print("Contains 'delayed':", "delayed" in review)
# See the output below
```

```python
# Create a list containing different types of data
#  Example-4 Classifying Data Programatically by Type
samples = [
    {"type": "structured", "value": {"Name": "Raj", "Age": 20}},
    {"type": "semi-structured", "value": {"tags": ["python", "data"]}},
    {"type": "unstructured", "value": "A long free-text customer comment..."}
]

# Loop through each item in the list
for s in samples:

    # Print the type of data and its Python data type
    print(s["type"], ":", type(s["value"]).__name__)
 # See the ouput below
```
